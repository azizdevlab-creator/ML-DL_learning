```python
import torch
import numpy as np
```

```python
print(torch.__version__)
print(f'cpu: {torch.cpu.device_count()}')
print(f'cuda: {torch.cuda.device_count()}')
print(torch.cuda.is_available())
```

```python
np.set_printoptions(precision=3)
a = [1, 2, 3]
b = np.array([4.4558, 5, 6], dtype = float)
print(b)
t_a = torch.tensor(a)
t_b = torch.from_numpy(b)
print(f't_a: {t_a}')
print(f't_b: {t_b}')
```

```python
torch.random.manual_seed(42)
t = torch.rand(2, 3)
print(f't: {t}')
print(f'shape of t: {t.shape}')
t_tr = torch.transpose(t, 0, 1)
print(f't_tr: {t_tr}')
print(f'shape os t_tr: {t_tr.shape}')
```

```python
torch.random.manual_seed(42)
t_dim3 = torch.rand(2, 3, 2)
print(f't_dim3: {t_dim3}')
print(f'shape of t_dim3: {t_dim3.shape}')
t_dim3_tr = torch.transpose(t_dim3, 1, 2)
print(f't_dim3_tr: {t_dim3_tr}')
print(f'shape os t_dim3_tr: {t_dim3_tr.shape}')
```

```python
t_ones5x7 = torch.ones(5, 7)
print(f'(5, 7) ones tensor: {t_ones5x7}')
t2x3 = torch.rand(2, 3)
print(f't2x3: {t2x3}')
print(f't2x3 like zeros tensor: {torch.zeros_like(t2x3)}')
```

```python
t2x1x3x1 = torch.randint(1, 6, size = (2, 1, 3, 1))
print(f't2x1x3x1: {t2x1x3x1}')
t_squeezed = torch.squeeze(t2x1x3x1, (1, 3))
print(f'squeezed: {t_squeezed}')
```

```python
t_zeros = torch.zeros(30)
print(t_zeros.reshape(5, 6))
```

```python
torch.manual_seed(42)
t1 = 2 * torch.rand(4, 5) - 1
t2 = torch.normal(mean=0, std=1, size = (4, 5))
t_mul = torch.multiply(t1, t2)
t_matmul = torch.matmul(t1, torch.transpose(t2, 0, 1))
print(f't_mul: {t_mul}')
print(f't_matmul: {t_matmul}')
print(f't1_sum: {torch.sum(t1, axis=0)}')
print(f't1_std: {torch.std(t1, dim=1)}')
print(f't1_mean: {torch.mean(t1, axis=0)}')
```

```python
print(torch.linalg.norm(t1, ord=2, dim=1))
print(np.round(np.sqrt(np.sum(np.square(np.array(t1)), axis = 1)), 4))
```

```python
t_ones3 = torch.ones(3)
t_zeros2 = torch.zeros(2)
torch.cat((t_ones3, t_zeros2), axis=0)
```

```python
t2x3 = torch.rand(2)
t_3 = torch.rand(2)
torch.stack([t2x3, t_3], axis=1)
```

```python
t = torch.rand((2, 3), requires_grad = True)
t2 = torch.sum(torch.pow(3, t))
t2.backward()
print(t.grad)
t = torch.randint(1, 100, size = (6,))
splits = torch.chunk(t, 2)
print([item.numpy() for item in splits])
print(splits)
```

```python
t = torch.rand(5)
splits = torch.split(t, split_size_or_sections=[2, 3])
print(splits)
```
