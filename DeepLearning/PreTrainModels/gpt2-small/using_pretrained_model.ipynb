```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
```

```python
%%capture
model_name = "azizdevlab/gpt2-small-uzbek"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)
model.eval()
```

```python
model
```

```python
prompt = "Sen kimsan"
inputs = tokenizer(prompt, return_tensors="pt").to(device)

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_length=500,
        do_sample=True,
        temperature=0.9,
        top_k=50,
        top_p=0.95
    )

generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
```

```python
from transformers import GPT2LMHeadModel, GPT2Config, GPT2Tokenizer
```

```python
config = GPT2Config()
model2 = GPT2LMHeadModel(config)
tokenizer2 = GPT2Tokenizer.from_pretrained("gpt2")
```

```python
tokenizer2.pad_token = tokenizer2.eos_token
tokenizer2.pad_token_id = tokenizer2.eos_token_id
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

```python
model2.to(device)
```

```python
prompt = "Sen kimsan"
inputs = tokenizer2(prompt, return_tensors="pt", add_special_tokens=True).to(device)
```

```python
with torch.no_grad():
    outputs2 = model2.generate(
        **inputs,
        max_length=500,
        do_sample=True,
        temperature=0.9,
        top_k=50,
        top_p=0.95
    )

generated_text = tokenizer2.decode(outputs2[0], skip_special_tokens=True)
```
