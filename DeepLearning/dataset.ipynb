```python
from torch.utils.data import DataLoader, Dataset, TensorDataset
import torch
import numpy as np
```

```python
t = torch.arange(8, dtype=torch.float32)
data_loader = DataLoader(t)
for item in data_loader:
    print(item)
```

```python
data_loader_batch = DataLoader(t, batch_size=3, drop_last=True)
for i, item in enumerate(data_loader_batch, 1):
    print(f'{i}: {item}')
```

```python
class JoinedData(Dataset):
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __len__(self):
        return len(self.x)
    
    def __getitem__(self, idx):
        return self.x[idx], self.y[idx]
    
t_x = torch.rand(4, 3)
t_y = torch.arange(4)

joined_data = JoinedData(t_x, t_y)
joined_data2 = TensorDataset(t_x, t_y)

print(f'len1: {joined_data.__len__()}')
print(f'len2: {joined_data2.__len__()}')

for item1, item2 in zip(joined_data, joined_data2):
    print(item1, item2)
```

```python
epoches = 2
data_loader = DataLoader(dataset=joined_data, batch_size=2, shuffle=True)
for epoche in range(epoches):
    print(f'epoche {epoche + 1}')
    for item in data_loader:
        print(f'x: {item[0]}, y: {item[1]}')
```
