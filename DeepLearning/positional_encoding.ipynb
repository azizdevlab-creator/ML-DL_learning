```python
import re
```

```python
with open("the-verdict.txt", 'rt') as file:
    raw_data = file.read()
```

```python
preprocessed = re.split(r"([.,:;!?_()'\"]|--|\s)", raw_data)
preprocessed = list(filter(str.strip, preprocessed))
unique_words = set(preprocessed)
vocab_size = len(unique_words)
vocab = {token:id for id, token in enumerate(unique_words)}
```

```python
class SimpleTokenizerV1:
    def __init__(self, vocab):
        self.stoid = vocab
        self.idtos = {id:s for s, id in vocab.items()}
    def encode(self, text):
        text = re.split(r"([.,:;!?()_'\"]|--|\s+)", text)
        print(text[:20])
        text = list(filter(str.strip, text))
        # text = [item for item in text if item.strip()]
        print(text)
        encoded = [self.stoid[token] for token in text]
        return encoded
    def decode(self, ids):
        text = " ".join([self.idtos[id] for id in ids])
        print(text)
        text = re.sub(r"\s+([.,:;!?()_'\"])", r"\1", text)
        return text
```

```python
tokenizer = SimpleTokenizerV1(vocab)
str1 = "head?   muddling,yet:yellow sensitive"
str2 = "head yet yellow sensitive"
token = tokenizer.encode(str1)
print(token)
text = tokenizer.decode(token)
print(text)
print(str1)
```

```python
unique_words = sorted(set(preprocessed))
unique_words.extend(["<|unk|>", "<|endoftext|>"])
vocab = {token:id for id, token in enumerate(unique_words)}
```

```python
class SimpleTokenizerV2:
    def __init__(self, vocab):
        self.stoid = vocab
        self.idtos = {id:s for s, id in vocab.items()}
    def encode(self, text):
        preprocessed = re.split(r"([.,:;!?()_'\"]|--|\s+)", text)
        preprocessed = list(filter(str.strip, preprocessed))
        print(preprocessed)
        preprocessed = [item if item in self.stoid
                        else "<|unk|>" for item in preprocessed]
        encoded = [self.stoid[token] for token in preprocessed]
        return encoded
    def decode(self, ids):
        text = " ".join([self.idtos[id] for id in ids])
        print(text)
        text = re.sub(r"\s+([.,:;!?()_'\"])", r"\1", text)
        return text
```

```python
text1 = "Hello, do you like tea?"
text2 = "In the sunlit terraces of the palace."
text = " <|endoftext|> ".join((text1, text2))
print(text)
```

```python
tokenizer2 = SimpleTokenizerV2(vocab)
encoded = tokenizer2.encode(text)
print(encoded)
```

```python
text = tokenizer2.decode(encoded)
```
