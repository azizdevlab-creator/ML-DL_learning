```python
import numpy as np
```

```python
def func(x):
    return 0.1 * x**2 - np.sin(x) + 5.

def model(x, w):
    return x @ w.T

coord_x = np.arange(-5.0, 5.0, 0.1)
coord_y = func(coord_x)

sz = len(coord_x)
eta = np.array([0.1, 0.01, 0.001, 0.0001])
w = np.array([0., 0., 0., 0.])
N = 200
X = np.c_[np.ones_like(coord_x), coord_x, coord_x**2, coord_x**3]
```

```python
for i in range(N):
    dw= (2/X.shape[0]) * (X @ w.T - coord_y) @ X
    w -= eta * dw
```

```python
Q = (1/X.shape[0]) * np.sum((model(X, w) - coord_y) ** 2)
Q
```
