```python
import numpy as np
```

```python
# Task 1
x_test = np.array([(9, 6), (2, 4), (-3, -1), (3, -2), (-3, 6), (7, -3), (6, 2)])
x = np.hstack((np.ones(x_test.shape[0]).reshape(-1, 1), x_test))
w = np.array([-14, 7, 5])
predict = np.sign(x @ w.T)
predict
```
