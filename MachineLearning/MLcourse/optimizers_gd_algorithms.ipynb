```python
import numpy as np
```

```python
# Momentum
def func(x):
    return -0.5 * x + 0.2 * x ** 2 - 0.01 * x ** 3 - 0.3 * np.sin(4*x)

def df(x):
    return -0.5 + 0.4*x - 0.03*x**2 - 1.2 * np.cos(4*x)

eta = 0.1
N = 200
lm = 0.8
v = 0
x = -3.5

for i in range(N):
    v = lm*v + (1-lm)*eta*df(x)
    x = x - v 
```

```python
# Nesterov Accelerated Gradient
def func(x):
    return 0.4 * x + 0.1 * np.sin(2*x) + 0.2 * np.cos(3*x)

def df(x):
    return 0.4 + 0.2*np.cos(2*x) - 0.6*np.sin(3*x)

eta = 1
N = 500
lm = 0.7
v = 0
x = 4

for _ in range(N):
    v = lm * v + (1 - lm) * eta * df(x - lm*v)
    x = x - v
```
