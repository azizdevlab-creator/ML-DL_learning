```python
np.random.seed(0)
x = np.arange(-1.0, 1.0, 0.1)

size_train = len(x)
w = [0.5, -0.3]
model_a = lambda m_x, m_w: (m_w[1] * m_x + m_w[0])
loss = lambda ax, y: (ax - y) ** 2

y = model_a(x, w) + np.random.normal(0, 0.1, len(x))
```
