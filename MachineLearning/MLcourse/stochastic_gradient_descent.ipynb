```python
import numpy as np
```

```python
def func(x):
    return 0.5 * x ** 2 - 0.1 / np.exp(-x) + 0.5 * np.cos(2 * x) - 2
def loss_fn(y_pred, y_true):
    return (y_pred - y_true)**2
coord_x = np.arange(-5.0, 5.0, 0.1)
coord_y = func(coord_x)
X = np.c_[np.ones_like(coord_x), coord_x, coord_x**2, np.cos(2*coord_x), np.sin(2*coord_x)] # X.shape = (100, 5)
sz = len(coord_x)
eta = np.array([0.01, 0.001, 0.0001, 0.01, 0.01])
w = np.array([0., 0., 0., 0., 0.])
N = 500
lm = 0.02

Qe = np.mean((X @ w - coord_y)**2)
np.random.seed(0)
for i in range(N):
    k = np.random.randint(0, sz-1)
    loss_k = (w.T @ X[k] - coord_y[k])**2
    dw = 2*(w.T @ X[k] - coord_y[k]) * X[k].T
    w = w - eta * dw
    Qe = lm * loss_k + (1 - lm)*Qe
Q = np.mean((X @ w - coord_y)**2)
```
